```python
import sys

# Aumentamos el límite de recursividad porque la lista ordenada
# creará una rama de 10,000 nodos de profundidad.
sys.setrecursionlimit(15000)

class NodoEstudiante:
    def __init__(self, id_matricula, nombre, edad, promedio):
        self.id_matricula = id_matricula
        self.nombre = nombre
        self.edad = edad
        self.promedio = promedio
        self.izquierdo = None
        self.derecho = None

class ArbolABB:
    def __init__(self):
        self.raiz = None

    # 1. Buscar un estudiante por ID
    def buscar(self, id_matricula):
        return self._buscar_recursivo(self.raiz, id_matricula)

    def _buscar_recursivo(self, nodo, id_matricula):
        if nodo is None or nodo.id_matricula == id_matricula:
            return nodo
        if id_matricula < nodo.id_matricula:
            return self._buscar_recursivo(nodo.izquierdo, id_matricula)
        return self._buscar_recursivo(nodo.derecho, id_matricula)

    # 2. Insertar nuevos estudiantes
    def insertar(self, id_matricula, nombre, edad, promedio):
        if self.raiz is None:
            self.raiz = NodoEstudiante(id_matricula, nombre, edad, promedio)
        else:
            self._insertar_recursivo(self.raiz, id_matricula, nombre, edad, promedio)

    def _insertar_recursivo(self, nodo, id_matricula, nombre, edad, promedio):
        if id_matricula < nodo.id_matricula:
            if nodo.izquierdo is None:
                nodo.izquierdo = NodoEstudiante(id_matricula, nombre, edad, promedio)
            else:
                self._insertar_recursivo(nodo.izquierdo, id_matricula, nombre, edad, promedio)
        elif id_matricula > nodo.id_matricula:
            if nodo.derecho is None:
                nodo.derecho = NodoEstudiante(id_matricula, nombre, edad, promedio)
            else:
                self._insertar_recursivo(nodo.derecho, id_matricula, nombre, edad, promedio)
        else:
            pass # Ignoramos IDs duplicados

    # 3. Listar todos los estudiantes en orden por ID
    def listar_en_orden(self):
        estudiantes = []
        self._inorden_recursivo(self.raiz, estudiantes)
        return estudiantes

    def _inorden_recursivo(self, nodo, lista):
        if nodo is not None:
            self._inorden_recursivo(nodo.izquierdo, lista)
            lista.append((nodo.id_matricula, nodo.nombre, nodo.edad, nodo.promedio))
            self._inorden_recursivo(nodo.derecho, lista)
```

```python
import random

# Cantidad total de registros
NUM_ESTUDIANTES = 10000

print(f"Generando lista de {NUM_ESTUDIANTES} estudiantes...")

# 1. Crear la lista ORDENADA
datos_ordenados = []

# Usamos IDs secuenciales (ej. del 100,000 al 109,999) para asegurar el orden
for i in range(NUM_ESTUDIANTES):
    id_matricula = 100000 + i
    nombre = f"Estudiante_{id_matricula}"
    edad = random.randint(17, 30)
    promedio = round(random.uniform(3.0, 5.0), 2)

    # Guardamos cada estudiante como una tupla
    datos_ordenados.append((id_matricula, nombre, edad, promedio))

# 2. Crear la lista ALEATORIA (Copia desordenada)
# Hacemos una copia superficial de la lista original
datos_aleatorios = datos_ordenados.copy()

# Desordenamos la copia
random.shuffle(datos_aleatorios)

print("¡Listas generadas con éxito!")
print(f"- Primer elemento de la lista ordenada: {datos_ordenados[0]}")
print(f"- Primer elemento de la lista aleatoria: {datos_aleatorios[0]}")
```
