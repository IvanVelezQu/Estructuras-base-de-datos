```python
class SistemaLista:
    def __init__(self):
        # Nuestra "base de datos" es una simple lista vacía
        self.estudiantes = []

    # 2. Insertar nuevos estudiantes
    def insertar(self, id_matricula, nombre, edad, promedio):
        # En una lista estándar, simplemente agregamos al final.
        # Esto es extremadamente rápido: O(1)
        nuevo_estudiante = (id_matricula, nombre, edad, promedio)
        self.estudiantes.append(nuevo_estudiante)

    # 1. Buscar un estudiante por ID
    def buscar(self, id_matricula):
        # Búsqueda Lineal: Tenemos que revisar elemento por elemento.
        # En el peor de los casos, se revisa los 10,000 registros: O(n)
        for est in self.estudiantes:
            if est[0] == id_matricula:  # est[0] es el id_matricula
                return est
        return None  # Si termina el ciclo y no lo encuentra

    # 3. Listar todos los estudiantes en orden por ID
    def listar_en_orden(self):
        # Timsort: Python usa un algoritmo de ordenamiento ultra optimizado.
        # Ordenamos basándonos en el primer elemento de la tupla (el ID).
        # Complejidad: O(n log n)
        return sorted(self.estudiantes, key=lambda x: x[0])
```

```python
import random
import time

print("Generando 10,000 estudiantes en RAM...")

# Creamos una lista ordenada
datos_ordenados = []
for i in range(10000):
    id_matricula = 100000 + i
    nombre = f"Estudiante_{id_matricula}"
    edad = random.randint(17, 30)
    promedio = round(random.uniform(3.0, 5.0), 2)
    datos_ordenados.append((id_matricula, nombre, edad, promedio))

# Creamos la versión aleatoria
datos_aleatorios = datos_ordenados.copy()
random.shuffle(datos_aleatorios)

print("¡Datos cargados en RAM listos para usar!\n")


# 2. Pruebas de tiempo

# Instanciamos las dos listas (una para datos ordenados y otra para aleatorios)
sistema_lista_ordenada = SistemaLista()
sistema_lista_aleatoria = SistemaLista()

# Tiempos de insercion
print("--- Tiempos de Inserción (Lista Nativa) ---")

# Inserción Aleatoria
inicio_ins_al = time.time()
for est in datos_aleatorios:
    sistema_lista_aleatoria.insertar(est[0], est[1], est[2], est[3])
fin_ins_al = time.time()
print(f"Inserción Aleatoria: {fin_ins_al - inicio_ins_al:.6f} segundos")

# Inserción Ordenada
inicio_ins_ord = time.time()
for est in datos_ordenados:
    sistema_lista_ordenada.insertar(est[0], est[1], est[2], est[3])
fin_ins_ord = time.time()
print(f"Inserción Ordenada: {fin_ins_ord - inicio_ins_ord:.6f} segundos")

# --- TIEMPOS DE BÚSQUEDA (LECTURA) ---
print("\n--- Tiempos de Búsqueda (Lista Nativa) ---")
# Extraemos los IDs para buscarlos
ids_a_buscar = [est[0] for est in datos_aleatorios]

# Búsqueda
inicio_busqueda = time.time()
for id_buscar in ids_a_buscar:
    # Solo buscamos en una, el tiempo será similar porque ambas hacen búsqueda lineal O(n)
    sistema_lista_aleatoria.buscar(id_buscar)
fin_busqueda = time.time()
print(f"Búsqueda de 10,000 IDs: {fin_busqueda - inicio_busqueda:.4f} segundos")
```
